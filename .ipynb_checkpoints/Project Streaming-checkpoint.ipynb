## 1. Digital Twin State Definition

The digital twin state represents the physical and operational health of the photodetector asset, synthesized continuously from ingested primitive telemetry signals (`temperature_c`, `optical_power_w`) combined with derived optoelectronic physics computations.

### State Schema Table

| State Field | Data Type | Units / Format | Primary Source | System Function / Role |
| :--- | :--- | :--- | :--- | :--- |
| `device_id` | `string` | Asset ID (`pd_sensor_01`) | Ingestion Stream (ZMQ / MQTT) | Primary key for physical asset telemetry |
| `temperature_c` | `float` | °C (24.5 - 25.5 °C nominal) | Ingestion Stream (ZMQ / MQTT) | Thermal sensor reading driving exponential dark current drift |
| `bias_voltage_v` | `float` | Volts (5.0 V baseline) | Ingestion Default | Ingested baseline bias voltage setting |
| `optical_power_w` | `float` | Watts (1 mW nominal) | Ingestion Stream (ZMQ / MQTT) | Ingested optical input signal power |
| `dark_current_a` | `float` | Amperes | Physics Engine (`evaluate_state`) | Derived thermal leakage current ($I_d = I_{d0} \cdot e^{0.07(T - T_0)}$) |
| `responsivity_a_w`| `float` | A/W | Physics Engine (`evaluate_state`) | Dynamic optical-to-electrical conversion efficiency ($R$) |
| `photocurrent_a` | `float` | Amperes | Physics Engine (`evaluate_state`) | Active signal current generated by incident light ($I_{ph} = P_{opt} \cdot R$) |
| `snr_db` | `float` | dB | Physics Engine (`evaluate_state`) | Real-time Signal-to-Noise Ratio calculated against shot noise |
| `health_index_pct`| `float` | Percentage (0 - 100%) | Physics Engine (`evaluate_state`) | Asset health score inversely proportional to dark current drift |
| `is_saturated` | `boolean` | `True` / `False` | Physics Engine (`evaluate_state`) | Safety condition indicator ($I_{total} \ge 10\text{ mA}$ limit) |
| `recommended_gain_mode`| `string` | `low_noise` \| `balanced` \| `high_sensitivity` | Behavioral AI Planner | Autonomous gain operational recommendation |
| `assigned_bias_v` | `float` | Volts | AI Command Planner | Feedback-controlled bias setting dispatched via MQTT to hardware |

## 2. Shared Physics Engine Verification

To ensure zero state drift between this evaluation notebook and the deployed microservices, we import `PhotodetectorPhysicsEngine` directly from the production repository path (`services.twin_engine.physics_model`).

```python
import sys
import os
import json

# Ensure twin_engine service package is in Python PATH
sys.path.append(os.path.abspath("services/twin_engine"))

from physics_model import PhotodetectorPhysicsEngine

# Instantiate production physics engine instance
physics_engine = PhotodetectorPhysicsEngine()
sample_state = physics_engine.evaluate_state(temp_c=25.0, bias_v=5.0, optical_power_w=0.001)

print("Physics Engine Imported Successfully!")
print("Nominal State Output (25.0 C, 5.0 V, 1 mW):")
print(json.dumps(sample_state, indent=2))
```

## 3. Real-Time Streaming Ingestion Acquisition (ZeroMQ / MQTT Sockets)

The code below connects directly to host ports exposed by your running containers:
- **ZMQ SUB Socket**: Connects to `tcp://localhost:5555` to capture raw telemetry frames.
- **MQTT Client**: Subscribes to `telemetry/pd_sensor_01` on `localhost:1883`.

```python
import zmq
import json

# Connect live ZMQ socket to running ingestion_service
context = zmq.Context()
zmq_socket = context.socket(zmq.SUB)
zmq_socket.connect("tcp://localhost:5555")
zmq_socket.setsockopt_string(zmq.SUBSCRIBE, "telemetry")

print("[ZMQ Socket] Connected to tcp://localhost:5555. Listening for live stream...")

captured_frames = []
for i in range(3):
    raw_msg = zmq_socket.recv_string()
    topic, payload_str = raw_msg.split(" ", 1)
    payload = json.loads(payload_str)
    
    # Compute state using shared physics engine
    state = physics_engine.evaluate_state(
        temp_c=payload["temperature_c"],
        bias_v=payload["bias_voltage_v"],
        optical_power_w=payload["optical_power_w"]
    )
    
    print(f"\n--- [Captured Stream Packet {i+1}] ---")
    print(f"Topic Prefix : {topic}")
    print(f"Device ID    : {payload['device_id']}")
    print(f"Temperature  : {payload['temperature_c']:.2f} °C")
    print(f"Optical Power: {payload['optical_power_w']:.6f} W")
    print(f"Timestamp    : {payload['timestamp']}")
    print(f"Derived SNR  : {state['snr_db']:.2f} dB")
    print(f"Health Index : {state['health_index_pct']:.2f}%")
    captured_frames.append((payload, state))

zmq_socket.close()
context.term()
print("\nStream Acquisition Test Successful! Captured 3 socket frames.")
```

## 4. InfluxDB Time-Series Query Verification (Persistence Proof)

This cell queries live persisted telemetry points directly from the running `influxdb_service` instance on port `8086`, proving that streaming telemetry and digital twin state calculations are being stored to disk.

```python
from influxdb_client import InfluxDBClient

INFLUX_URL = "http://localhost:8086"
INFLUX_TOKEN = "my-super-secret-auth-token"
INFLUX_ORG = "opto-twin"
INFLUX_BUCKET = "photodetector_telemetry"

client = InfluxDBClient(url=INFLUX_URL, token=INFLUX_TOKEN, org=INFLUX_ORG)
query_api = client.query_api()

flux_query = f'''
from(bucket: "{INFLUX_BUCKET}")
  |> range(start: -5m)
  |> filter(fn: (r) => r["_measurement"] == "photodetector_state")
  |> pivot(rowKey:["_time"], columnKey: ["_field"], valueColumn: "_value")
  |> limit(n: 5)
'''

print(f"[InfluxDB Client] Successfully connected to {INFLUX_URL}!")
print(f"Querying bucket '{INFLUX_BUCKET}' for stored measurements...\n")

tables = query_api.query(flux_query)

print(f"{'Timestamp':<30} {'Device ID':<14} {'Temp (°C)':<10} {'Bias (V)':<9} {'SNR (dB)':<9} {'Health (%)':<10}")
print("-" * 86)

for table in tables:
    for record in table.records:
        ts = str(record.get_time())
        device = record.values.get("device_id", "pd_sensor_01")
        temp = record.values.get("temperature_c", 0.0)
        bias = record.values.get("bias_voltage_v", 0.0)
        snr = record.values.get("snr_db", 0.0)
        health = record.values.get("health_index_pct", 0.0)
        print(f"{ts:<30} {device:<14} {temp:<10.2f} {bias:<9.2f} {snr:<9.2f} {health:<10.2f}")

client.close()
print("\nPersistence Proof Complete! Verified records stored in InfluxDB volume.")
```
