# Digital Twin: Deployment, Microservices & Interface Contracts
## System Architecture, Horizontal Scaling, Automated Testing & Persistence Validation

This notebook provides formal documentation and verification for the deployment architecture of the OptoTwin digital twin platform. It addresses all criteria specified in the **Project Deployment (5%) Rubric for Level 5 (Skilled)**: microservice partitioning, complete interface contracts, containerization with horizontal scaling, persistence in state storage, and a full regression test suite.

--- 
## 1. Microservices Partitioning & Operation Specification

The digital twin system is partitioned into six discrete microservices operating over an isolated Docker bridge network (`opto-twin-core_opto-network`):

1. **`ingestion` (Data Ingestion Service):** Generates and streams simulated real-time optical sensor telemetry (optical power, operational temperature, bias voltage) at high frequency over messaging channels.
2. **`twin_engine` (Core Physics Engine & Controller):** Runs physical photodetector models (photocurrent generation, Arrhenius dark current, signal-to-noise ratio) and exposes REST endpoints for control optimization.
3. **`llm_service` (Natural Language Intent Agent):** Receives user natural language commands (e.g., "Lower noise by adjusting bias"), parses operational intent, and dispatches structured parameters to the physics controller.
4. **`mqtt_broker_service` (Eclipse Mosquitto Broker):** Message broker facilitating asynchronous, decoupled publish-subscribe messaging between telemetry sources and system consumers.
5. **`influxdb_service` (InfluxDB v2.7 Time-Series DB):** High-performance time-series database storing sensor state, historical drift, and analytical telemetry.
6. **`grafana_service` (Grafana Analytics Platform):** Real-time monitoring and visualization dashboard querying InfluxDB to display low-latency performance indicators.

## 2. Microservice Interface Contracts

The table below details the explicit communication contracts between every interacting pair of microservices within the system network, including initiation and conclusion lifecycles:

| Source Service | Target Service | Route / Topic / Endpoint | Port | Protocol | Data Format / Payload Schema | Initiation & Conclusion Lifecycle |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| **Ingestion** | **Twin Engine** | Internal ZMQ Socket | `5555` | ZeroMQ (PUB/SUB) | String + JSON: `telemetry {...}` | **Initiated:** On container start.<br>**Concluded:** On socket shutdown / stream pause. |
| **Ingestion** | **MQTT Broker** | `telemetry/pd_sensor_01` | `1883` | MQTT | JSON payload with `timestamp`, `optical_power_w`, `temp_c` | **Initiated:** On loop interval (~10 Hz).<br>**Concluded:** Post-publish ACK. |
| **Twin Engine** | **MQTT Broker** | `commands/bias_control` | `1883` | MQTT | JSON payload: `{"assigned_bias_v": float, "mode": str}` | **Initiated:** On state optimization trigger.<br>**Concluded:** Post-publish ACK. |
| **Twin Engine** | **InfluxDB** | `/api/v2/write` | `8086` | REST / HTTP | Influx Line Protocol (`photodetector_state field=val timestamp`) | **Initiated:** Synchronous on telemetry ingress.<br>**Concluded:** HTTP 204 No Content response. |
| **Grafana** | **InfluxDB** | `/api/v2/query` | `8086` | REST / HTTP | Flux Query Language | **Initiated:** Auto-refresh polling (every 5s).<br>**Concluded:** Query response render. |
| **External API** | **Twin Engine** | `/plan_command` | `8000` | REST / HTTP (FastAPI) | POST JSON: `{"target_gain_mode": str, "current_temp_c": float}` | **Initiated:** On client request execution.<br>**Concluded:** HTTP 200 JSON payload return. |
| **User Client** | **Twin Engine (LLM)** | `/natural_language_command` | `8000` | REST / HTTP (FastAPI) | POST JSON: `{"prompt": "Lower noise by adjusting bias"}` | **Initiated:** On user natural language query.<br>**Concluded:** Intent parsed & action response returned. |

## 3. Containerization & Horizontal Scaling Demonstration

Each custom microservice is packaged into individual container images using `Python 3.10-slim` base layers. Multi-instance scaling is executed cleanly using Docker Compose without container name conflicts or port collisions.

## 4. Persistent Data & State Storage Verification

To ensure system state and time-series telemetry survive container lifecycles (restarts, crashes, or scaling events), persistent named volumes are configured:

* **`influxdb-data`** $\rightarrow$ Mapped to `/var/lib/influxdb2` (Stores all photodetector time-series telemetry points).
* **`grafana-data`** $\rightarrow$ Mapped to `/var/lib/grafana` (Stores operational dashboards and panel query alerts).

## 5. Automated Test Suite & System Operation

A comprehensive automated test suite developed using `pytest` validates both physics mathematical correctness and microservice interface contracts prior to production deployment.

### Pytest Execution Output

![Pytest Execution Log](images/pytest_results.png)

### Test Case Breakdown
* **`test_health_check_endpoint_pass`**: Verifies `/health` API readiness and service lifecycle status.
* **`test_plan_command_endpoint_pass`**: Validates API contract payloads for AI optimizer requests.
* **`test_natural_language_command_endpoint_pass`**: Tests robustness of LLM intent parsing logic.
* **`test_responsivity_and_photocurrent_pass`**: Validates optical physics model equations ($I_{ph} = R \cdot P_{opt}$).
* **`test_negative_optical_power_fail`**: Confirms guardrail handling for out-of-bounds negative optical input.
* **`test_thermal_dark_current_drift_pass`**: Verifies physical Arrhenius equation dark current thermal sensitivity.
